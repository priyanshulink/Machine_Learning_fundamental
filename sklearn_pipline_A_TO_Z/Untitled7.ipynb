```python
import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import MinMaxScaler
from sklearn.tree import DecisionTreeClassifier
```

```python
df = pd.read_csv('/content/train.csv')
```

```python
df.head()
```

```python
df = df.drop(columns=['PassengerId','Name','Ticket','Cabin'],axis=1)
```

```python
from os import X_OK
from re import X
# step1- Train_Test_Split
X_train,X_test,y_train,y_test = train_test_split(df.drop(columns=['Survived'],axis=1),
                                               df['Survived'],
                                               test_size=0.2,
                                               random_state=42)
```

```python
df.isnull().sum()
```

```python
# apply simpleImputer on age column
si_age = SimpleImputer()
si_embarked = SimpleImputer(strategy='most_frequent')

X_train_age = si_age.fit_transform(X_train[['Age']])
X_train_embarked = si_embarked.fit_transform(X_train[['Embarked']])

X_test_age = si_age.fit_transform(X_test[['Age']])
X_test_embarked = si_embarked.fit_transform(X_test[['Embarked']])
```

```python
X_train_age.shape
```

```python
X_train_embarked
```

```python
# One Hot Encoding on sex and embarked
ohe_sex = OneHotEncoder(sparse_output = False, handle_unknown = 'ignore')
ohe_embarked = OneHotEncoder(sparse_output = False, handle_unknown = 'ignore')

X_train_sex = ohe_sex.fit_transform(X_train[['Sex']])
X_train_embarked = ohe_embarked.fit_transform(X_train_embarked)

X_test_sex = ohe_sex.fit_transform(X_test[['Sex']])
X_test_embarked = ohe_embarked.fit_transform(X_test_embarked)
```

```python
X_train_sex

```

```python
X_train_embarked
```

```python
# Creating new numpy array and drop these three column 'Sex','Age','Embarked'
X_train_rem = X_train.drop(columns=['Sex','Age','Embarked'],axis=1)
X_test_rem = X_test.drop(columns=['Sex','Age','Embarked'],axis=1)
```

```python
X_train_transformed = np.concatenate((X_train_rem,X_train_age,X_train_sex,X_train_embarked),axis=1)
X_test_transformed = np.concatenate((X_test_rem,X_test_age,X_test_sex,X_test_embarked),axis=1)
```

```python
 # apply Decision tree
 clf = DecisionTreeClassifier()
 clf.fit(X_train_transformed,y_train)
```

```python
y_pred = clf.predict(X_test_transformed)
```

```python
y_pred
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_test,y_pred)
```
