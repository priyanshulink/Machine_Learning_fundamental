```python
import pandas as pd
import seaborn as sns
```

# importing four different data set

```python
tips = sns.load_dataset('tips')
titanic = sns.load_dataset('titanic')
flights = sns.load_dataset('flights')
iris = sns.load_dataset('iris')
```

```python
tips.head()
```

# 1. Scatterplot (Numerical- Numerical)

```python
sns.scatterplot(x=tips['total_bill'],y=tips['tip'])
```

```python
# Multivariate analysis - find which customer is mail and femail
sns.scatterplot(x=tips['total_bill'],y=tips['tip'],hue=tips['sex'],style=tips['smoker'],size=tips['size'])
```

# Bar plot (Numberical + categorical )

```python
titanic.head()
```

```python
sns.barplot(x=titanic['pclass'],y=titanic['fare'], hue=titanic['sex'])
```

# Box Plot (Numerical - Categorical)

```python
sns.boxplot(x=titanic['sex'],y=titanic['age'], hue=titanic['survived'])
```

# Distplot (numerical - Categorical)

```python
sns.displot(
    data=titanic,
    x='age',
    hue='survived',
    kind='kde'
)
```

# HeatMap (Categorical - Categorical)

```python
titanic.head(3)
```

```python
sns.heatmap(pd.crosstab(titanic['pclass'],titanic['survived']))
```

```python
(titanic.groupby('pclass')['survived'].mean()*100).plot(kind='bar')
```

# ClusterMap (Categorical - Categorical)

```python
sns.clustermap(pd.crosstab(titanic['sibsp'],titanic['survived']))
```

# Pariplot
this automatically detect all numerical cols and plot scatter plot of each other

```python
iris.head()
```

```python
sns.pairplot(iris,hue='species')
```

# Line Plot (Numerical - Numerical)

this is used when x axis data is change on time

```python
flights.head()
```

```python
new = flights.groupby('year')['passengers'].sum().reset_index()
```

```python
sns.lineplot(x=new['year'],y=new['passengers'])
```
