```python
import numpy as np
import pandas as pd
```

```python
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import MinMaxScaler
from sklearn.pipeline import Pipeline,make_pipeline
from sklearn.feature_selection import SelectKBest, chi2
from sklearn.tree import DecisionTreeClassifier
```

```python
df = pd.read_csv('train.csv')
```

```python
df.head(5)
```

Titanic Survival Prediction Workflow

Input Data
    ↓
1. Handle Missing Values
   - Age and Embarked contain missing values.
   - Use a ColumnTransformer with SimpleImputer to fill them.

    ↓
2. One-Hot Encoding
   - Convert the Sex and Embarked categorical columns into numerical columns.

    ↓
3. Feature Scaling
   - Scale numerical features so they are in a similar numerical range.

    ↓
4. Feature Selection
   - Select the five best features using SelectKBest and chi-square (chi2).

    ↓
5. Model Training
   - Train a Decision Tree Classifier using the selected features.

    ↓
Output
   - Predict whether a passenger survived or did not survive.

```python
df.drop(columns=['PassengerId','Name','Ticket','Cabin'],inplace=True)
```

```python
# step 1 --> train/test/split

X_train, X_test, y_train, y_test = train_test_split(df.drop(columns=['Survived']),
                                                   df['Survived'],
                                                   test_size=0.2,
                                                   random_state=42)
```

```python
X_train.head(5)
```

```python
y_train.head(5)
```

```python
# Imputation transformer
trf1 = ColumnTransformer([
    ('impute_age',SimpleImputer(),[2]), # [2] --> this is index value of age
    ('impute_embarked',SimpleImputer(strategy='most_frequent'), [6]) # [6] --> this is index value of embarked 
], remainder='passthrough')
```

```python
# One Hot Encoding

trf2 = ColumnTransformer([
    ('ohe_sex_embarked', OneHotEncoder(sparse_output=False, handle_unknown='ignore'),[1,6])
],remainder='passthrough')

# know we have 10 columns
```

```python
# Scalling --> MinMaxScaler() -->> i use this becouse we further use "feature scaling"

trf3 = ColumnTransformer([
    ('scale', MinMaxScaler(),slice(0,10)) # here i apply minmaxscaler on all colum on catogerical column tooo.
])
```

```python
# feture selection 
trf4 = SelectKBest(score_func=chi2,k=8) # k = 8 mean top8 most importent function used
```

```python
# train the model
trf5 = DecisionTreeClassifier()
```

# creating pipeline

```python
pipe = Pipeline([
    ('trf1' ,trf1),
    ('trf2' ,trf2),
    ('trf3' ,trf3),
    ('trf4' ,trf4),
    ('trf5' ,trf5),
])
```

# Pipeline Vs make_pipeline
pipeline required naming steps, make_pipeline does not.

(some applies to column transformer vs make_column_transformer)

```python
# alternate syntex
pipe = make_pipeline(trf1,trf2,trf3,trf4,trf5)
```

```python
#train
pipe.fit(X_train,y_train)
```

# Explore the Pipeline

```python
# code here
pipe.named_steps # this will tell you what particular setps that pipeline fallow 
```

```python
# let calculate mean value of simple imputer
pipe.named_steps['columntransformer-1'].transformers_[0][1].statistics_
```

```python
# let us see most frequent value 
pipe.named_steps['columntransformer-1'].transformers_[1][1].statistics_
```

```python
# Display pipleline
from sklearn import set_config
set_config(display='diagram')
```

```python
# predict
y_pred = pipe.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_test,y_pred)
```

# cross validation using pipeline


```python
# cross validation using cross_val_score
from sklearn.model_selection import cross_val_score
cross_val_score(pipe, X_train, y_train, cv = 5, scoring='accuracy').mean()
```

# GridSearch using pipeline 

```python
# hyper tuining --> means kisi bhi algorithem me uske seating ko change kar ke uske accuracy ko change kar sakte ho
```

```python
# gridsearchcv
params = {
    'selectkbest__k': [1, 2, 3, 4, 5, 6, 7, 8],
    'decisiontreeclassifier__max_depth': [1, 2, 3, 4, 5]
}
```

```python
from sklearn.model_selection import GridSearchCV
grid = GridSearchCV(pipe, params, cv=5, scoring='accuracy')
grid.fit(X_train, y_train)
```
