```python
import  numpy as np
import pandas as pd
```

```python
df = pd.read_csv("titanic.csv")
```

```python
df.head(5)
```

```python
df['number'].unique()
```

```python
import matplotlib.pyplot as plt

df['number'].value_counts().plot.bar()
plt.title("Passenger Traveling With")
plt.show()
```

```python
# extract numerical part
df['number_numerical'] = pd.to_numeric(df['number'],errors='coerce',downcast='integer')
```

```python
# extract categorical part
df['number_categorical'] = np.where(df['number_numerical'].isnull(),df['number'],np.nan)
```

```python
df.head()
```

```python
df['Cabin'].unique()
```

```python
df['Ticket'].unique()
```

```python
df['Cabin_num']= df['Cabin'].str.extract('(\d+)') # string.extract kiye aur regular expression send kara jo ke numerica part ko extract karta hai 
df['Cabin_cat']= df['Cabin'].str[0] # capture the first letter
df.head()
```

```python
df['Cabin_cat'].value_counts().plot(kind='bar')
```

# for ticket column

```python
#extract the last bit of ticket as number
df['Ticket_num'] = df['Ticket'].apply(lambda s: s.split()[-1])
df['Ticket_num'] = pd.to_numeric(df['Ticket_num'],
                                errors='coerce',
                                downcast='integer')
# extract the first part of ticket as category
df['Ticket_cat'] = df['Ticket'].apply(lambda s: s.split()[0])
df['Ticket_cat'] = np.where(df['Ticket_cat'].str.isdigit(), np.nan,
                           df['Ticket_cat'])
```
