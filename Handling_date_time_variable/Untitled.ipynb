# working with Date

```python
#Converting to datatype
date['date'] = pd.to_datetime(date['date'])
```

```python
date.info()
```

# Extract Year 

```python
date['date_year'] = date['date'].dt.year
date.head()
```

# Extract months

```python
date['date_month'] = date['date'].dt.month
date.head()
```

```python
date['date_month'] = date['date'].dt.month_name()
date.head()
```

# Extract day

```python
date["date_day"] = date['date'].dt.day
date.head()
```

# Extract day of week

```python
date['date_ofweek'] = date['date'].dt.dayofweek
date.head()
```

```python
# day of week _name
date['date_dow_name'] = date['date'].dt.day_name()
date.head()
```

```python
date.drop(columns=['product_id', 'city_id', 'orders'], inplace=True)

date.head()
```

# is weekend?

```python
date['date_is_weekend'] = np.where(date['date_dow_name'].isin(['Sunday','Saturday']),0,1)
date.head()
```

```python
# extract week of year
date['date_week'] = date['date'].dt.isocalendar().week
date.head()
```

```python
# extract quarter

date['Quarter']=date['date'].dt.quarter
date.head()
```

```python
# extract semester
date['semester'] = np.where(date['Quarter'].isin([1,2]),1,2)
date.head()
```

```python
import datetime

today = datetime.datetime.today()

today
```

```python
today-date['date']
```

```python
# if you only need days
(today-date['date']).dt.days
```

```python
# month passed
np.round((today - date['date']).dt.days / 30, 0)
```
